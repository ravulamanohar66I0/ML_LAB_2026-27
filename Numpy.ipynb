Version

```python
import numpy as np

print(np.__version__)
```

1D array

```python

a = np.array([10, 20, 30, 40, 50])

print(a)
```

2D array

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(data)
```

NumPy properties

```python
print(data.shape)
print(data.ndim)
print(data.size)
print(data.dtype)
```

Indexing

```python
print(a[0])
print(a[2])
print(a[-1])
```

Slicing

```python
print(a[1:4])
```

```python
print(a[:2])
print(data[1:])
```

Mathematical operations

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

Statistical functions

```python
data = np.array([10, 20, 30, 40, 50])

print(np.mean(data))
print(np.median(data))
print(np.std(data))
print(np.var(data))
print(np.min(data))
print(np.max(data))
print(np.sum(data))
```

Working with missing values


```python
data = np.array([10, 20, np.nan, 40, 50])

print(np.mean(data))
```

```python
print(np.nanmean(data))
print(np.nansum(data))
print(np.nanmin(data))
print(np.nanmax(data))
print(np.isnan(data))
```

 np.concatenate

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.concatenate((a, b))

print(result)
```

Reshape

```python
a = np.arange(1, 7)

print(a.reshape(2, 3))
```

Boolean filtering

```python
data = np.array([10, 25, 30, 45, 50])

print(data[data > 30])
```
